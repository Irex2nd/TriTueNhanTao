## Bai13.py

```python
def tron_mang_tong(a, b):
    ketqua = []
    i = j = 0
    while i < len(a) or j < len(b):
        if i < len(a) and j < len(b):
            ketqua.append(a[i] + b[j])
            i += 1
            j += 1
        elif i < len(a):
            ketqua.append(a[i])
            i += 1
        else:
            ketqua.append(b[j])
            j += 1
    return ketqua
```

## Bai14.py

```python
import numpy as np
import random

# a. Tạo mảng ngẫu nhiên
m, n = 4, 5
a = np.random.randint(-20, 50, size=(m, n))
print("Mảng a:\n", a)

# b. Xuất dòng k
k = 2
print(f"Dòng {k}:", a[k])

# c. Xuất cột k
print(f"Cột {k}:", a[:, k])

# d. Dòng có tổng lớn nhất
dong_max = np.argmax(np.sum(a, axis=1))
print(f"Dòng có tổng lớn nhất (dòng {dong_max}):", a[dong_max])

# e. Cột có tích nhỏ nhất
tich_cot = np.prod(a, axis=0)
cot_min = np.argmin(tich_cot)
print(f"Cột có tích nhỏ nhất (cột {cot_min}):", a[:, cot_min])

# f. Phần tử dòng chẵn, cột lẻ
print("Phần tử dòng chẵn, cột lẻ:")
for i in range(m):
    if i % 2 == 0:
        for j in range(n):
            if j % 2 == 1:
                print(a[i,j], end=' ')
print()

# g. Trung bình phần tử chẵn thuộc dòng lẻ
chan_dong_le = [a[i,j] for i in range(m) if i % 2 == 1 for j in range(n) if a[i,j] % 2 == 0]
print("Trung bình chẵn dòng lẻ:", np.mean(chan_dong_le) if chan_dong_le else 0)

# h. Trung bình các phần tử thuộc biên
bien = np.concatenate((a[0], a[-1], a[:,0], a[:,-1]))
print("Trung bình biên:", np.mean(bien))

# i. Trung bình tích các phần tử không thuộc biên
noi = a[1:-1, 1:-1]
print("Trung bình tích nội:", np.mean(noi) if noi.size > 0 else 0)
```

## Bai15.py

```python
class SINHVIEN:
    def __init__(self, masv, ten, namsinh, dtb):
        self.masv = masv
        self.ten = ten
        self.namsinh = namsinh
        self.dtb = dtb

# Tạo danh sách sinh viên mẫu
ds_sv = [
    SINHVIEN("01DH0001", "Nguyễn Văn A", 2005, 7.5),
    SINHVIEN("02DH0002", "Trần Thị Lan", 2004, 4.8),
    SINHVIEN("03DH0003", "Lê Văn B", 2006, 9.2),
    SINHVIEN("04CN0004", "Phạm Thị Lan", 2005, 6.0),
    SINHVIEN("05DH0005", "Hoàng Thị C", 2003, 5.5)
]

# c. Số sinh viên đủ điều kiện lên lớp (dtb >= 5)
def dem_len_lop(ds):
    return sum(1 for sv in ds if sv.dtb >= 5)

# d. Xuất sinh viên đủ 20 tuổi (giả sử năm hiện tại 2025)
def sv_du_20_tuoi(ds):
    nam_hien_tai = 2025
    return [sv for sv in ds if nam_hien_tai - sv.namsinh >= 20]

# e. Đếm sinh viên hệ Đại học (chứa "DH" ở vị trí 2,3)
def dem_he_dh(ds):
    return sum(1 for sv in ds if len(sv.masv) >= 4 and sv.masv[2:4] == "DH")

# f. Đếm sinh viên tên "Lan"
def dem_ten_lan(ds):
    return sum(1 for sv in ds if sv.ten.find("Lan") != -1)

# g. Đếm sinh viên họ "Phan"
def dem_ho_phan(ds):
    return sum(1 for sv in ds if sv.ten.startswith("Phan"))

# Thực thi
print("Số SV đủ lên lớp:", dem_len_lop(ds_sv))
print("SV đủ 20 tuổi:", [sv.ten for sv in sv_du_20_tuoi(ds_sv)])
print("Số SV hệ DH:", dem_he_dh(ds_sv))
print("Số SV tên Lan:", dem_ten_lan(ds_sv))
print("Số SV họ Phan:", dem_ho_phan(ds_sv))
```

## btvn1.py

```python
def tron_mang_min(a, b):
    ketqua = []
    i = j = 0
    while i < len(a) or j < len(b):
        if i < len(a) and j < len(b):
            ketqua.append(min(a[i], b[j]))
            i += 1
            j += 1
        elif i < len(a):
            ketqua.append(a[i])
            i += 1
        else:
            ketqua.append(b[j])
            j += 1
    return ketqua
```

## btvn2.py

```python
import numpy as np

def bai2_tam_giac_tren_cheo_phu(a):
    n = a.shape[0]
    tong = 0
    for i in range(n):
        for j in range(n):
            if i + j <= n - 1:   # Tam giác trên của đường chéo phụ
                tong += a[i, j]
    return tong

def bai2_chuyen_am_thanh_tuyet_doi(a):
    return np.abs(a)

def bai2_thay_chan_bang_x(a, x):
    a_copy = a.copy()
    a_copy[a_copy % 2 == 0] = x
    return a_copy

def bai2_toan_chan(a):
    return np.all(a % 2 == 0)

def bai2_doi_xung(a):
    return np.all(a == a.T)

def bai2_cheo_chinh_tang_dan(a):
    return np.all(np.diff(np.diag(a)) > 0)

def bai2_tam_giac_duoi_cheo_phu(a):
    n = a.shape[0]
    for i in range(n):
        for j in range(n):
            if i + j >= n - 1:
                print(a[i, j], end=' ')
        print()
```
